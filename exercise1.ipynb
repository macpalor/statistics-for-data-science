## 1.
Let $G_1$ be the probability that the first side is green and $G_2$ the probability that the other side is green. Only one of the three cards has green on both sides, thus $P(G_1 \cap G_2) = \frac{1}{3}$. Since we have three cards and each of them has two sides, we have 6 possible sides that we can see initially. Since 3 of the 6 initial possibilites are green, we have $P(G_1)=\frac{3}{6}=\frac{1}{2}$. From the definition of conditional probability we then obtain
$$
P(G_2 \mid G_1) = \frac{P(G_1 \cap G_2)}{P(G_1)} = \frac{\frac{1}{3}}{\frac{1}{2}}=\frac{2}{3}.
$$

## 2.
A family has 4 different ways of having two children: $\{BB, GG, BG, GB\}$, where $B$ is a boy and $G$ is a girl. Since we know that at least one the children is a girl, the outcome $BB$ is not possible. This leaves only 3 outcomes $\{GG, BG, GB\}$ and out of these, two include a boy. Thus $P(G \cap B) = \frac{2}{3}$.
