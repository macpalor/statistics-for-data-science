# Problem 1
## 1.
The mean of Bernoulli distribution is $p$, thus the plug-in estimator for $p$ is the sample mean, i.e.
$$
\hat{p} = \frac{1}{n}\sum_{i=1}^n X_i=\bar{X}.
$$
## 2.
The standard error is 
$$
\begin{align*}
\text{se}(\hat{p}) &= \sqrt{\text{Var}(\hat{p})} \\
&= \sqrt{\text{Var}(\bar{X})} \\
&= \sqrt{\text{Var} \left( \frac{1}{n}\sum_{i=1}^n X_i \right)} \\
&= \sqrt{\frac{1}{n^2} \sum_{i=1}^n \text{Var}(X_i)} \\
&= \sqrt{\frac{1}{n^2} np(1-p)} \\
&= \sqrt{\frac{p(1-p)}{n}},
\end{align*}
$$
where we used the fact that the variance of Bernoulli distribution is $p(1-p)$, $\text{Var}(aX)=a^2 \text{Var}(X)$ and for i.i.d. variables $\text{Var}(X_1 + \dots + X_n)= \text{Var}(X_1) + \dots + \text{Var}(X_n)$. Since the parameter $p$ is unknown, we use the plug-in estimator for that and thus obtain
$$
\text{se}(\hat{p}) = \sqrt{\frac{\hat{p}(1-\hat{p})}{n}}.
$$

# Problem 2

```python
import pandas as pd
import numpy as np
# import dataset
data = pd.read_csv("./data/week2_heart_disease_data.csv", index_col=0)

chol = data['chol']
heart_disease = 1 * (data['heart-disease'] != 0)
```

## 1.

```python
corr = chol.corr(heart_disease)
print(f"corr = {corr:.4f}")
```

## 2.

```python
rng = np.random.default_rng(42)

B = 10000 # N of bootstrap samples
N = len(chol) # sample size
corrs = np.zeros(B)
for i in range(B):
    idx = rng.choice(N, N, replace=True)
    chol_boot = chol[idx]
    heart_disease_boot = heart_disease[idx]
    corrs[i] = chol_boot.corr(heart_disease_boot)

se = np.std(corrs, ddof=1)
print(f"Standard error = {se:.4f}")
    
```

## 3.

```python
import scipy.stats as stats

alpha = 0.05
z_alpha = stats.norm.ppf(1 - alpha/2)
normal_lower = corr - z_alpha * se
normal_upper = corr + z_alpha * se

print(f"95% Normal interval: ({normal_lower:.3f}, {normal_upper:.3f})")

percentile_lower = np.percentile(corrs, 100 * alpha/2)
percentile_upper = np.percentile(corrs, 100 * (1 - alpha/2))

print(f"95% Percentile interval: ({percentile_lower:.3f}, {percentile_upper:.3f})")

pivotal_lower = 2 * corr - np.percentile(corrs, 100 * (1 - alpha/2))
pivotal_upper = 2 * corr - np.percentile(corrs, 100 * alpha/2)

print(f"95% Pivotal interval: ({pivotal_lower:.3f}, {pivotal_upper:.3f})")
```

Let $Y=\sum_{i=1}^{n}X_i = n\bar{X}$. For i.i.d variables $E(n\bar{X})=n E(\bar{X}) = \sum_{i=1}^n E(X_i) = n\lambda=100$ and $\text{Var}(n\bar{X})= n^2 \text{Var}(\bar{X})=\sum_{i=100}^n \text{Var}(X_i)= n\lambda=100$, where we used the fact the mean and variance are the same for Poisson distributed variables. Then by the central limit theorem $Y \sim \mathcal{N}(100, 100)$ for large $n$. Then
$$
\mathbb{P}(Y < 90) = \Phi \left( \frac{90 - 100}{10} \right)=\Phi(-1) \approx 0.1587.
$$

# Problem 4
Let $X_1,\dots , X_n \sim \text{Poisson}(\lambda)$ and $\hat{\lambda}=\frac{1}{n}\sum_{i=100}^n X_i$.

## 1.
Since $X_i$ are i.i.d., the bias is
$$
\begin{align*}
\text{bias}(\hat{\lambda}) &= E(\hat{\lambda}) -  \lambda \\
&= \frac{1}{n}\sum_{i=1}^n E(X_i) - \lambda \\
&= \frac{1}{n} n\lambda - \lambda \\
&= 0.
\end{align*}
$$

## 2.
The standard error is
$$
\begin{align*}
\text{se}(\hat{\lambda}) &= \sqrt{\text{Var}(\hat{\lambda})} \\
&= \sqrt{\frac{1}{n^2}\sum_{i=1}^n \text{Var}(X_i)} \\
&= \sqrt{\frac{1}{n^2}n\lambda} \\
&= \sqrt{\frac{\lambda}{n}}
\end{align*}.
$$

## 3.
Using the bias-variance decomposition, we obtain
$$
\text{MSE}(\hat{\lambda}) = \text{bias}^2(\hat{\lambda}) + \text{Var}(\hat{\lambda}) = \frac{\lambda}{n}.
$$

```python
m = 1000
rng = np.random.default_rng(42)
p_arr = np.zeros(m)
for i in range(m):
    samples = rng.binomial(10, 0.5, size=100)
    p = (samples == 5).mean()
    p_arr[i] = p
fraction = ((p_arr >= 0.16) & (p_arr <= 0.33)).mean()
print(f"fraction of repeats = {fraction}")
```

# 3.
For $X\sim \text{Binom}(10, 1/2)$, $\text{Pr}(X=5) = \binom{10}{5}(1/2)^5(1-1/2)^5=\binom{10}{5}(1/2)^{10} = \frac{252}{1024} = \frac{63}{256} \equiv p$. Getting the result $\text{Pr}(X=5)$ based on 100 samples is itself a random variable $Z \sim \text{Binom}(100, p)$ and the estimator is then $Y=Z/100$. The probability is thus
$$
\begin{align*}
\text{Pr}(0.16 \leq Y \leq 0.33) &= \text{Pr}\left( 0.16 \leq \frac{Z}{100} \leq 0.33 \right) \\
&= \text{Pr}(16 \leq Z \leq 33) \\
&= \sum_{k=16}^{33} \binom{100}{k}p^k (1-p)^{100-k} \\
&= \sum_{k=16}^{33} \binom{100}{k}\left( \frac{63}{256} \right)^k \left(1- \frac{63}{256} \right)^{100-k} \\
&\approx 0.964.
\end{align*}
$$
