# Problem 1
## 1.
The PMF for the Bernoulli distribution is $P(X=x)=p^x(1-p)^{1-x}$, so the likelihood is
$$
\mathcal{L}(p)=\prod_{i=1}^N p^{X_i}(1-p)^{1-X_i}=p^k(1-p)^{N-k},
$$
where $k=\sum_{i=1}^N X_i$ is the number of successes. The log-likelihood is then
$$
\ell(p)=k \log p +(N-k) \log (1-p).
$$
Setting the derivative to zero we find
$$
\begin{align*}
\ell ' (p) &= \frac{k}{p} - \frac{N-k}{1-p} = 0 \\
\implies \frac{k}{p} &= \frac{N-k}{1-p} \\
\implies k-pk &= Np - pk \\
\implies \hat{p} &= \frac{k}{N}.
\end{align*}
$$
For $k=23$ and $N=100$ this yields $\hat{p}=23/100=0.23$.

## 2.
The Fisher information for a single Bernoulli trial is
$$
\begin{align*}
I(p) &= - \mathbb{E}_p \left( \frac{d^2}{dp^2} \log (p^x(1-p)^{1-x}) \right) \\
&= -\mathbb{E}_p \left( \frac{d^2}{dp^2} (x\log p +(1-x)\log(1-p)) \right) \\
&= -\mathbb{E}_p \left( \frac{d}{dp} \left( \frac{x}{p} - \frac{1-x}{1-p}\right) \right) \\
&= -\mathbb{E}_p \left( -\frac{x}{p^2} - \frac{1-x}{(1-p)^2} \right) \\
&= \frac{1}{p} + \frac{1}{1-p} \\
&= \frac{p + (1-p)}{p(1-p)} \\
&= \frac{1}{p(1-p)},
\end{align*}
$$
where we used the linearity of expectation and $\mathbb{E}_p (x)=p$.
For $N$ observations we have $I_n(p)= n I(p)=\frac{N}{p(1-p)}$. The standard error of $\hat{p}$ is 
$$
\hat{\text{se}} = \frac{1}{\sqrt{I_n(\hat{p})}} = \sqrt{\frac{\hat{p}(1-\hat{p})}{N}}.
$$
The confidence interval is given by
$$
\text{CI} = (\hat{p} - z_{\alpha/2}\, \hat{\text{se}},\, \hat{p} + z_{\alpha/2}\, \hat{\text{se}}).
$$
A 95\% confidence interval has $z_{0.025}\approx 1.96$, so 
$$
\text{95\% CI}= \hat{p} \pm 1.96 \, \hat{\text{se}}.
$$
Substituting $\hat{p}=0.23$ and $N=100$, we get
$$
\text{95\% CI} \approx (0.1475, 0.3125).
$$

## 3.

```python
import numpy as np

rng = np.random.default_rng(42)

k, N = 23, 100
p = k/N

data = np.array(k*[1] + (N-k)*[0])
B = 10000
p_boot = np.zeros(B)

for i in range(B):
    p_boot[i] = rng.choice(data, size=N, replace=True).mean()

# Percentile interval
alpha=0.05
percentile_lower = np.percentile(p_boot, 100 * alpha/2)
percentile_upper = np.percentile(p_boot, 100 * (1 - alpha/2))

print(f"95% Percentile interval: ({percentile_lower:.3f}, {percentile_upper:.3f})")

# Pivotal interval  
pivotal_lower = 2 * p - np.percentile(p_boot, 100 * (1 - alpha/2))
pivotal_upper = 2 * p - np.percentile(p_boot, 100 * alpha/2)

print(f"95% Pivotal interval: ({pivotal_lower:.3f}, {pivotal_upper:.3f})")
```

The PDF is
$$
f(x; a, b) = \begin{cases}
  \frac{1}{b-a}, & \text{for } a\leq x \leq b \\
   0, & \text{otherwise} 
\end{cases}.
$$
The likelihood is 
$$
\mathcal{L}(a,b) = \prod_{i=1}^4 \frac{1}{b-a} = \frac{1}{(b-a)^4}.
$$
In order to maximize the likelihood, the denominator $b-a$ must be as small as possible. This means that $a$ should be as large as possible, and $b$ as small as possible. However, in order to satisfy the domain of the PDF, $a$ can't be any larger than any of the $X_i$ and likewise $b$ can't be any smaller than any of the $X_i$. Thus $\hat{a}=\min (X_1, X_2, X_3, X_4)$ and $\hat{b}=\max (X_1, X_2, X_3, X_4)$.

We have $X_1, \dots,X_n \sim \text{Poisson}(\lambda)$. The first theoretical moment is $E_\lambda [X] =\lambda$. The first sample moment is $\bar{X}=\frac{1}{n} \sum_{i=1}^n X_i$. By equating them we get

\hat{\lambda}=\bar{X}=\frac{1}{n}\sum_{i=1}^n X_i.
$$

# Problem 4
The PMF of the Poisson distribution is 
$$
f(x;\lambda) = \frac{\lambda^x e^{-\lambda}}{x!}.
$$
The likelihood is
$$
\mathcal{L}(\lambda) = \prod_{i=1}^n \frac{\lambda^{X_i} e^{-\lambda}}{X_i!} = \frac{\lambda^{\sum_{i=1}^n X_i}e^{-n\lambda}}{\prod_{i=1}^n X_i!}
$$
and the log-likelihood is then
$$
\ell (\lambda) = \log \lambda \sum_{i=1}^n X_i - n\lambda - \sum_{i=1}^n \log (X_i!).
$$
Setting the derivative to zero and solving for $\lambda$ we find
$$
\begin{align*}
\ell '(\lambda) &= \frac{\sum_{i=1}^n X_i}{\lambda} - n = 0 \\
\implies \hat{\lambda} &= \frac{1}{n} \sum_{i=1}^n X_i. 
\end{align*}
$$

We can also check that the second derivative
$$
\ell ''(\lambda) = -\frac{\sum_{i=1}^n X_i}{\lambda^2} < 0, \quad \forall \lambda,
$$
thus we have found a maximum.

# Problem 5
The PDF for the inverse Gamma distribution is 
$$
f(x;a,b) = \frac{b^a}{\Gamma (a)}x^{-a-1}e^{-b/x},
$$
where $a,b > 0$. The likelihood is
$$
\mathcal{L}(a,b)=\prod_{i=1}^n \frac{b^a}{\Gamma (a)}x_i^{-a-1}e^{-b/x_i} = \left( \frac{b^a}{\Gamma (a)} \right)^n e^{-b \sum_i 1/x_i} \prod_i x_i^{-a-1}
$$
and the log likelihood is
$$
\ell(a,b) = n [a\log b - \log \Gamma(a)] -b \sum_i \frac{1}{x_i} - (a+1) \sum_i \log x_i.
$$
To maximize the log likelihood, we minimize its negative.

```python
from scipy.optimize import minimize
from scipy.special import gammaln

data = np.loadtxt("./data/week3_exercise5.txt", skiprows=1)
n = len(data)
sum1 = np.sum(1/data)
sum2 = np.sum(np.log(data))

def objective_func(theta):
    a, b = theta
    l = n*(a*np.log(b) - gammaln(a)) - b*sum1 - (a + 1)*sum2
    return -l # minimize the negative

x0 = [1, 1] # initial guess
bounds = ((0, None), (0, None)) # a, b > 0
result = minimize(objective_func, x0=x0, bounds=bounds, method="L-BFGS-B")
print(f"Optimization Success: {result.success}")
print(f"Optimal a, b values: {result.x}")
print(f"Minimum function value: {result.fun}")
```
